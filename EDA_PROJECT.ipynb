Dataframe Operations

• dataframe1 should contain the attributes: ProductID, NameofProd, and Price.
• dataframe2 should contain the attributes: ProductID and Company_Location.
• dataframe3 should contain the attributes: ProductID, NameofProd, and Price.
Insert a minimum of 10 records into each data frame.
Using the above data frames, perform the following operations:
1. Append the records of dataframe1 and dataframe3.
2. Concatenate the data frames wherever applicable.
3. Merge dataframe1 and dataframe2 using the ProductID attribute.
4. Perform the following join operations between dataframe1 and dataframe2:
o Inner Join
o Left Join
o Right Join
o Full Outer Join

```python
import pandas as pd
dataframe1 = pd.DataFrame({
    'ProductID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'NameofProd': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Headphones', 'USB Drive', 'Webcam', 'Printer', 'Router', 'Tablet'],
    'Price': [1200, 25, 45, 150, 80, 15, 60, 200, 75, 300]
})

dataframe2 = pd.DataFrame({
    'ProductID': [101, 102, 103, 104, 105, 111, 112, 113, 114, 115],
    'Company_Location': ['USA', 'China', 'Taiwan', 'Korea', 'Japan', 'Germany', 'India', 'Canada', 'UK', 'Vietnam']
})

dataframe3 = pd.DataFrame({
    'ProductID': [116, 117, 118, 119, 120, 121, 122, 123, 124, 125],
    'NameofProd': ['Smartwatch', 'Speaker', 'Microphone', 'Projector', 'Scanner', 'External HDD', 'Stylus', 'VR Headset', 'Graphics Card', 'Power Bank'],
    'Price': [250, 120, 90, 400, 180, 110, 30, 350, 500, 40]
})

print(" Original DataFrames")
print("DataFrame 1:\n", dataframe1.head(2))
print("\nDataFrame 2:\n", dataframe2.head(2))
print("\nDataFrame 3:\n", dataframe3.head(2))
```

Data Transformations

Write an program to create a numeric vector using your first name (Fname) as the object name. The vector
should contain 10 numeric values. The first value must be the last four digits of your Registration Number,
followed by any nine integer values of your choice.
Perform the following data transformation techniques on the vector:
1. Normalize the data to the following ranges:
o [-3, +3]
o [-1, +1]
o [0, 1] (Min-Max Normalization)
2. Transform the data into Z-score format.
3. Transform the data using Decimal Scaling Normalization

```python
import numpy as np
import pandas as pd

Thamarai = np.array([190, 45, 12, 88, 304, 250, 75, 115, 60, 410], dtype=float)
print("Original Vector:\n", Thamarai)

def min_max_scale(v, a, b):
    v_min = np.min(v)
    v_max = np.max(v)
    return a + ((v - v_min) * (b - a)) / (v_max - v_min)

norm_neg3_pos3 = min_max_scale(Thamarai, -3, 3)
norm_neg1_pos1 = min_max_scale(Thamarai, -1, 1)
norm_0_1 = min_max_scale(Thamarai, 0, 1)
```

```python
z_score = (Thamarai - np.mean(Thamarai)) / np.std(Thamarai)
max_val = np.max(np.abs(Thamarai))
j = np.ceil(np.log10(max_val))
decimal_scaling = Thamarai / (10**j)
```

```python
transformations_df = pd.DataFrame({
    'Original': Thamarai,
    'Range [-3, 3]': norm_neg3_pos3,
    'Range [-1, 1]': norm_neg1_pos1,
    'Min-Max [0, 1]': norm_0_1,
    'Z-Score': z_score,
    'Decimal Scaling': decimal_scaling
})

print("\n--- Data Transformation Results ---")
print(transformations_df.round(4))
```

```python
# 1. Load the built-in AirPassengers datase
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import statsmodels.api as sm

data = sm.datasets.get_rdataset("AirPassengers").data
passengers = pd.Series(
    data['value'].values,
    index=pd.date_range(start='1949-01-01', periods=len(data), freq='MS'),
    name='Passengers'
)
print(passengers.head())
```

```python
# 2. Explore the first five and last five observations of the dataset
print("First five observations:")
print(passengers.head())
print("\nLast five observations:")
print(passengers.tail())
```

```python
# 3. Identify the starting and ending time periods of the dataset. (start, end)
print("Starting period:", passengers.index[0])
print("Ending period:", passengers.index[-1])
```

```python
# 4. Determine the frequency and time-series characteristics of the dataset. (frequency, class)
print("Frequency: 12 observations per year")
print("Data type:", type(passengers))
print("Number of observations:", len(passengers))
print("Time interval:", passengers.index.freq)
print("Time series class: pandas Series with a DatetimeIndex")
print("\nSummary statistics:")
print(passengers.describe())
```

```python
# 5. Visualize the monthly passenger trend over the entire period. (plot)
plt.figure(figsize=(12, 5))
plt.plot(passengers, color='blue')
plt.title('Monthly Air Passengers (1949–1960)')
plt.xlabel('Year')
plt.ylabel('Number of Passengers (Thousands)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 6. Decompose the time series into trend, seasonal, and random components. (decompose)
decomposition = seasonal_decompose(
    passengers,
    model='additive',
    period=12
)

print("Time series decomposition completed.")
```

```python
# 7. Plot the decomposed components to examine the underlying patterns
decomposition.plot()
plt.suptitle('Air Passengers Time Series Decomposition', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# 8.Analyze the autocorrelation and partial autocorrelation of the time series. (acf, pacf)
fig, axes = plt.subplots(2, 1, figsize=(10, 8))
plot_acf(passengers, lags=40, ax=axes[0])
axes[0].set_title('Autocorrelation Function (ACF)')
plot_pacf(passengers, lags=40, ax=axes[1], method='ywm')
axes[1].set_title('Partial Autocorrelation Function (PACF)')
plt.tight_layout()
plt.show()
```

```python
# 9.  Fit an ARIMA model to the AirPassengers time series. (arima)
model = ARIMA(passengers, order=(2, 1, 2))
fitted_model = model.fit()

print("ARIMA model fitted successfully.")
```

```python
# 10.  Display and interpret the fitted ARIMA model parameters and model summary.(summary)
print(fitted_model.summary())
```

```python
# 11. Forecast the number of passengers for the next 12 months using the fitted ARIMA model.(predict)
forecast_result = fitted_model.get_forecast(steps=12)
forecast_values = forecast_result.predicted_mean
confidence_intervals = forecast_result.conf_int()
forecast_df = pd.DataFrame({
    'Forecast': forecast_values,
    'Lower 95% CI': confidence_intervals.iloc[:, 0],
    'Upper 95% CI': confidence_intervals.iloc[:, 1]
})
print("Forecast for the next 12 months:")
print(forecast_df.round(2))
```

```python
# 12. Visualize the actual and forecasted passenger values with appropriate labels and title. (plot, lines)
plt.figure(figsize=(12, 6))
plt.plot(
    passengers.index,
    passengers,
    label='Actual Passengers',
    color='blue'
)
plt.plot(
    forecast_values.index,
    forecast_values,
    label='Forecasted Passengers',
    color='red',
    linestyle='--',
    marker='o'
)
plt.fill_between(
    forecast_values.index,
    confidence_intervals.iloc[:, 0],
    confidence_intervals.iloc[:, 1],
    color='red',
    alpha=0.15,
    label='95% Confidence Interval'
)
plt.title('Air Passengers: Actual Values and 12-Month Forecast')
plt.xlabel('Year')
plt.ylabel('Number of Passengers (Thousands)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 1. Import libraries and load dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.impute import SimpleImputer
df = pd.read_csv('/content/prison.csv')

#2. view the dataset
display(df.head())
print("Dataset shape:", df.shape)
print("\nDataset information:")
df.info()
```

```python
# 3. Check missing values
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# 4. Check for the dependent variable distribution. (table)
print("Distribution of govelec:")
print(df['govelec'].value_counts(dropna=False))
```

```python
# 5. Select numerical features and exclude categorical codes
exclude_cols = ['state', 'govelec', 'final1', 'final2', 'rownames']
numeric_data = df.select_dtypes(include=np.number).drop(
    columns=exclude_cols,
    errors='ignore'
)
numeric_data = numeric_data.dropna(axis=1, how='all')
imputer = SimpleImputer(strategy='mean')
X_imputed = imputer.fit_transform(numeric_data)
print("Numerical features selected:")
print(numeric_data.columns.tolist())

# 6. Standardize the data(scale)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)
print("\nStandardized data (first 5 rows):")
display(pd.DataFrame(X_scaled, columns=numeric_data.columns).head())
```

```python
# 7. K-Means clustering with 2 clusters
kmeans2 = KMeans(n_clusters=2, random_state=42, n_init=10)
clusters2 = kmeans2.fit_predict(X_scaled)
print("K-Means clustering completed with k = 2")
```

```python
# 8. Display K-Means results
results = df.copy()
results['Cluster'] = clusters2

display(results.head(10))

print("\nNumber of samples in each cluster:")
print(results['Cluster'].value_counts().sort_index())
```

```python
# 9. Display cluster centers
centers_scaled = kmeans2.cluster_centers_
centers_original = scaler.inverse_transform(centers_scaled)
centers_df = pd.DataFrame(
    centers_original,
    columns=numeric_data.columns
)
print("Cluster Centers:")
display(centers_df.round(3))
```

```python
# 10. Calculate WCSS for K = 1 to 10
wcss = []
for k in range(1, 11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X_scaled)
    wcss.append(model.inertia_)

print("WCSS values:")
for k, value in enumerate(wcss, start=1):
    print(f"K = {k}: WCSS = {value:.2f}")
```

```python
# 11. Display the elbow plot.
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker='o', color='blue')
plt.title('Elbow Method for Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')
plt.xticks(range(1, 11))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 12. Apply K-means clustering based on the elbow recommendation
optimal_k = 3
optimal_model = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)
optimal_clusters = optimal_model.fit_predict(X_scaled)
results['Optimal_Cluster'] = optimal_clusters
print("Selected K:", optimal_k)
print("\nCluster counts:")
print(results['Optimal_Cluster'].value_counts().sort_index())
display(results.head(10))
```

```python
# 1. Load dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv('/content/prison.csv')

# 2. View dataset
display(df.head())
print("Dataset shape:", df.shape)
```

```python
# 3. Check missing values
print(df.isnull().sum())
```

```python
# 4. Select numerical attributes
exclude_cols = ['state', 'govelec', 'final1', 'final2', 'rownames']
numeric_data = df.select_dtypes(include=np.number).drop(
    columns=exclude_cols,
    errors='ignore'
)
numeric_data = numeric_data.dropna(axis=1, how='all')
imputer = SimpleImputer(strategy='mean')
X_imputed = imputer.fit_transform(numeric_data)

# 5. Standardize the data (scale)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)
print("Standardized numerical data:")
display(pd.DataFrame(X_scaled, columns=numeric_data.columns).head())
```

```python
# 6. Calculate Euclidean distance (dist)
from scipy.spatial.distance import pdist
sample_size = min(300, len(X_scaled))
rng = np.random.default_rng(42)
sample_indices = np.sort(
    rng.choice(len(X_scaled), size=sample_size, replace=False)
)
X_sample = X_scaled[sample_indices]
euclidean_distances = pdist(X_sample, metric='euclidean')
print("Number of sampled observations:", sample_size)
print("Number of pairwise distances:", len(euclidean_distances))
print("First 10 distances:", euclidean_distances[:10])
```

```python
# 7. Perform Hierarchical Clustering using Ward's method (hclust)
from scipy.cluster.hierarchy import linkage
linked = linkage(X_sample, method='ward', metric='euclidean')
print("Hierarchical clustering completed.")
print("Linkage matrix shape:", linked.shape)
```

```python
# 8. Plot dendrogram
from scipy.cluster.hierarchy import dendrogram
plt.figure(figsize=(14, 6))
dendrogram(
    linked,
    truncate_mode='lastp',
    p=30,
    leaf_rotation=90,
    leaf_font_size=9
)
plt.title("Hierarchical Clustering Dendrogram (Ward's Method)")
plt.xlabel('Cluster or Observation')
plt.ylabel('Euclidean Distance')
plt.tight_layout()
plt.show()
```

```python
# 9. Divide the data into 3 clusters and Display cluster assignments(cutree)
from scipy.cluster.hierarchy import fcluster
hierarchical_clusters = fcluster(
    linked,
    t=3,
    criterion='maxclust'
)
hierarchical_results = pd.DataFrame({
    'Original_Row_Index': sample_indices,
    'Cluster': hierarchical_clusters
})
print("Cluster assignments:")
display(hierarchical_results.head(20))
```

```python
# 10. Display number of samples in each cluster. (table)
cluster_counts = hierarchical_results['Cluster'].value_counts().sort_index()
print("Number of samples in each hierarchical cluster:")
print(cluster_counts)
```

```python
# 1. Load the dataset sent to your email into R, and verify that the packages ggplot2, ggfortify,
# FactoMineR, and factoextra are loaded.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

df = pd.read_csv('/content/prison.csv')

print("Dataset loaded successfully!")
print("Libraries loaded: pandas, numpy, matplotlib, seaborn, sklearn")
```

```python
# 2. Display the total number of rows and columns in the dataset.
print("Total number of rows:", df.shape[0])
print("Total number of columns:", df.shape[1])
```

```python
# 3. Extract only the numeric columns, clean the data, and normalize it using the scale()
# function into an object named data_scaled.

numeric_data = df.select_dtypes(include=[np.number]).copy()
numeric_data = numeric_data.drop(
    columns=['rownames', 'state', 'govelec', 'final1', 'final2'],
    errors='ignore'
)
numeric_data = numeric_data.replace([np.inf, -np.inf], np.nan)
numeric_data = numeric_data.dropna(axis=1, how='all')
imputer = SimpleImputer(strategy='mean')
numeric_clean = pd.DataFrame(
    imputer.fit_transform(numeric_data),
    columns=numeric_data.columns
)
scaler = StandardScaler()
data_scaled = pd.DataFrame(
    scaler.fit_transform(numeric_clean),
    columns=numeric_clean.columns
)
print("Scaled data:")
display(data_scaled.head())
```

```python
# 4. Perform Principal Component Analysis on data_scaled using either the PCA() or prcomp()
# function, and print its summary.
pca = PCA()
pca_result = pca.fit_transform(data_scaled)

print("PCA Summary")
print("Number of components:", pca.n_components_)
print("Explained variance:")
print(pca.explained_variance_)
print("\nPercentage of variance explained:")
print(pca.explained_variance_ratio_ * 100)
print("\nCumulative percentage of variance:")
print(np.cumsum(pca.explained_variance_ratio_) * 100)
```

```python
# 5. Plot a scree plot showing the percentage of variance for each component with data labels,
# setting the y-axis limit from 0 to 60.

variance_percent = pca.explained_variance_ratio_ * 100
components = np.arange(1, len(variance_percent) + 1)
plt.figure(figsize=(10, 6))
plt.plot(components, variance_percent, marker='o')
plt.xticks(components)
plt.ylim(0, 60)
plt.xlabel('Principal Component')
plt.ylabel('Percentage of Variance Explained')
plt.title('Scree Plot')
plt.grid(True)

for i, value in enumerate(variance_percent):
    if value <= 60:
        plt.text(i + 1, value, f'{value:.2f}%', ha='center', va='bottom')

plt.show()
```

```python
# 6. Create a PCA biplot with non-overlapping labels and a proper title.
scores = pca_result[:, :2]
plt.figure(figsize=(12, 8))
plt.scatter(scores[:, 0], scores[:, 1], alpha=0.7)
loadings = pca.components_.T[:, :2]

for i, feature in enumerate(data_scaled.columns):
    plt.arrow(
        0, 0,
        loadings[i, 0] * 3,
        loadings[i, 1] * 3,
        color='red',
        alpha=0.6,
        head_width=0.08,
        length_includes_head=True
    )
    plt.text(
        loadings[i, 0] * 3.2,
        loadings[i, 1] * 3.2,
        feature,
        color='red',
        fontsize=8
    )

plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('PCA Biplot')
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True)
plt.show()
```

```python
# 7. Extract the PCA scores into a data frame named pca_scores and display the first six rows.
pca_scores = pd.DataFrame(
    pca_result,
    columns=[f'PC{i+1}' for i in range(pca_result.shape[1])]
)
display(pca_scores.head(6))
```

```python
# 8. Draw a scatter plot of the first two principal components (PC1 vs. PC2) with blue circular
# points and your registration number in the title

plt.figure(figsize=(8, 6))
plt.scatter(
    pca_scores['PC1'],
    pca_scores['PC2'],
    color='blue',
    marker='o'
)
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('PCA Scatter Plot - Registration Number: 23BDS0190')
plt.grid(True)
plt.show()
```
